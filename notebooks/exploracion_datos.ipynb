## Filtrado, preparacion e integracion de datos

```python
#Renombrar columnas a Snake Case

def to_snake_case(df):
    df.columns = (
        df.columns.str.strip()
                  .str.lower()
                  .str.replace(" ", "_")
                  .str.replace("-", "_")
                  .str.replace(r"[^\w_]", "", regex=True)
    )
    return df

# Aplicar a ambos dataframes
recorridos = to_snake_case(recorridos)
usuarios = to_snake_case(usuarios)

# Verificar
print("Columnas recorridos:")
print(recorridos.columns.tolist())

print("\nColumnas usuarios:")
print(usuarios.columns.tolist())
```

```python
# Función para intentar convertir columnas con fechas a datetime
def convertir_fechas(df):
    for col in df.columns:
        if "fecha" in col or "date" in col:
            try:
                df[col] = pd.to_datetime(df[col], errors="coerce", dayfirst=True)
                print(f"Columna convertida a datetime: {col}")
            except Exception as e:
                print(f"No se pudo convertir {col}: {e}")
    return df

# Aplicar a los dos datasets
recorridos = convertir_fechas(recorridos)
usuarios = convertir_fechas(usuarios)

# Verificar tipos de datos actualizados
print("\n=== dtypes recorridos ===")
print(recorridos.dtypes)

print("\n=== dtypes usuarios ===")
print(usuarios.dtypes)
```

```python
col_fecha = "fecha_origen_recorrido"  # o "fecha_destino_recorrido" si preferís

# Filtro para incluir todo 31/09 (hasta 23:59:59):
inicio = pd.Timestamp(2024, 7, 1)
fin_exclusivo = pd.Timestamp(2024, 10, 1)  # < 1/10/2024 equivale a <= 30/9/2024 23:59:59

mask = (recorridos[col_fecha] >= inicio) & (recorridos[col_fecha] < fin_exclusivo)
viajes_jul_ago = recorridos.loc[mask].copy()

print(f"Cantidad de viajes entre {inicio.date()} y 2024-09-30 (inclusive): {len(viajes_jul_ago)}")

# Desglose por mes
viajes_jul_ago["mes"] = viajes_jul_ago[col_fecha].dt.to_period("M")
print("\nViajes por mes:")
print(viajes_jul_ago["mes"].value_counts().sort_index())
```

Aca guardo en data/interim


```python
df_completo = viajes_jul_ago.merge(
    usuarios,
    on="id_usuario",
    how="left"
)
```

```python
# Revisar columnas disponibles en ambos
print("Columnas recorridos:", recorridos.columns.tolist())
print("Columnas usuarios:", usuarios.columns.tolist())

# Verificar tipos de dato de id_usuario
print("\nTipo en recorridos:", recorridos["id_usuario"].dtype)
print("Tipo en usuarios:", usuarios["id_usuario"].dtype)

# Chequear valores únicos
print("\nCantidad de id_usuario distintos en recorridos:", recorridos["id_usuario"].nunique())
print("Cantidad de id_usuario distintos en usuarios:", usuarios["id_usuario"].nunique())

# ¿Hay ids en recorridos que no estén en usuarios?
ids_no_en_usuarios = set(recorridos["id_usuario"].dropna().unique()) - set(usuarios["id_usuario"].unique())
print(f"\nIDs en recorridos que no están en usuarios: {len(ids_no_en_usuarios)}")

# ¿Y viceversa?
ids_no_en_recorridos = set(usuarios["id_usuario"].unique()) - set(recorridos["id_usuario"].dropna().unique())
print(f"IDs en usuarios que no están en recorridos: {len(ids_no_en_recorridos)}")
```

```python
# Asegurarse de que id_usuario sea del mismo tipo
viajes_jul_ago["id_usuario"] = viajes_jul_ago["id_usuario"].astype("Int64")
usuarios["id_usuario"] = usuarios["id_usuario"].astype("Int64")

# Realizar el join (merge)
df_completo = viajes_jul_ago.merge(
    usuarios,
    on="id_usuario",
    how="left"  # left join para mantener todos los recorridos
)

# Verificar resultado
print("Dimensiones del DataFrame resultante:", df_completo.shape)
display(df_completo.head())
```

```python
# Calcular duración en minutos
df_completo["duracion_viaje_minutos"] = (
    (df_completo["fecha_destino_recorrido"] - df_completo["fecha_origen_recorrido"])
    .dt.total_seconds() / 60
)

# Verificar algunas filas
display(df_completo[["fecha_origen_recorrido", "fecha_destino_recorrido", "duracion_viaje_minutos"]].head())
```
